```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
```

```python
df=pd.read_csv("profit_sales_analysis_dataset.csv")
```

```python
df
```

```python
sns.histplot(df['Profit'],bins=7)
plt.title("Distribution of Profit")
plt.show()
```

```python
sns.histplot(df['Sales'],bins=10)
plt.title("Distribution of Sales")
plt.show()
```

```python
sns.boxplot(x=df['Profit'])
plt.title("Dboxplot of Profit")
plt.show() 
```

```python
sns.boxplot(x=df['Sales'])
plt.title("Dboxplot of Sales")
plt.show() 
```

```python
sns.barplot(x="Quantity",y="Profit_Margin_Percent",data=df)
plt.title("Quantity vsProfit Margin")
plt.show()
```

```python
sns.scatterplot(x="Quantity",y="Profit_Margin_Percent",data=df)
plt.title("Quantity vsProfit Margin")
plt.show()
```

```python
numeric_cols=["Unit_Price","Quantity","Discount_Percent","Sales","Profit","Profit_Margin_Percent","Cost"]
```

```python
joshna=df[numeric_cols].corr()
```

```python
joshna
```

```python
sns.heatmap(joshna,
            annot=True,
            cmap="coolwarm",
            cbar=False,
            square=True,
            linewidth=1)
plt.title("profit_margin")
plt.show()
```
