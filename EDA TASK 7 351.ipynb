```python
Categorical_cols=["gender","race/ethnicity","parental level of education","lunch","test preparation course"]
```

```python
Categorical_cols
```

```python
for cols in Categorical_cols:
    df[cols]=df[cols].astype(str).str.strip().str.lower()
```

```python
df[cols]
```

```python
score_cols=["math score","reading score","writing score"]
```

```python
score_cols
```

```python
for cols in score_cols:
    mean=df[cols].mean()
    print("mean score:",mean)
    median=df[cols].median()
    print("median score:",median)
    mode=df[cols].mode()
    print("mode score:",mode)
    std=df[cols].std()
    print("std score:",std)
    Q1=df[cols].quantile(0.25)
    print("Q1 score:",Q1)
    Q3=df[cols].quantile(0.75)
    print("Q3 score:",Q3)
```

```python
print(df[score_cols].describe())
```

```python
df["Total Marks"]=df[score_cols].sum(axis=1)
```

```python
df["Percentage"]=(df["Total Marks"]/300)*100
```

```python
print(df[score_cols+["Total Marks","Percentage"]])
```

```python
performance=[]
for Percentage in df["Percentage"]:
    if Percentage>=90:
        performance.append("excellent")
    elif Percentage>=75:
        performance.append("very good")
    elif Percentage>=60:
        performance.append("good")
    elif Percentage>=50:
        performance.append("average")
    else:
        performance.append("need a improvment")
df["performance"]=performance
print(df[["Percentage","performance"]])
        
```

```python
for col in["math score","reading score","writing score"]:
    
    Q1=df[col].quantile(0.25)
    Q3=df[col].quantile(0.75)
    
    IQR=Q3-Q1
    
    lower=Q1-1.5*IQR
    upper=Q3+1.5*IQR
    
    outliers=df[
        (df[col]<lower)|
        (df[col]>upper)
        ]

    print(outliers[col])
    
```
