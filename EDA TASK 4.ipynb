```python
import pandas as pd
df=pd.read_csv("stock_data_50_records.csv")
```

```python
df
```

```python
df.info()
```

```python
df["Date"]=pd.to_datetime(df["Date"])
df.dtypes
```

```python
df=df.sort_values("Date")
```

```python
df
```

```python
df["MA_20"]=df["Close"].rolling(20).mean()
df["MA_50"]=df["Close"].rolling(50).mean()
df
```

```python
df["Daily_Return"]=(df["Close"]/df["Open"])*100
df
```

```python
mean_return=df["Daily_Return"].mean()
print("mean_return",mean_return)

variance_return=df["Daily_Return"].var()
print("variance_return",variance_return)

standard_return=df["Daily_Return"].std()
print("standard_return",standard_return)
```

```python
import matplotlib.pyplot as plt
```

```python
plt.figure(figsize=(12,6))
plt
plt.plot(df["Date"],df["Open"],label="Open")
plt.plot(df["Date"],df["Close"],label="Close")
plt.plot(df["Date"],df["High"],label="High")
plt.plot(df["Date"],df["Low"],label="Low")
plt.title("OCHL price over time")
plt.xlabel("Date")
plt.ylabel("Price")
plt.legend()
plt.xticks(rotation=45)
plt.show()
```

```python
plt.figure(figsize=(12,6))
plt
plt.plot(df["Date"],df["Volume"],label="Volume")
plt.title("trading volume prices over time")
plt.xlabel("Date")
plt.ylabel("volume")
plt.legend()
plt.xticks(rotation=45)
plt.show()
```

```python
plt.figure(figsize=(12,6))
plt
plt.plot(df["Date"],df["MA_20"],label="MA_20")
plt.plot(df["Date"],df["Close"],label="Close")
plt.plot(df["Date"],df["MA_50"],label="MA_50")
plt.title("moving average prices over time")
plt.xlabel("Date")
plt.ylabel("Price")
plt.legend()
plt.xticks(rotation=45)
plt.show() 
```

```python
import seaborn as sns
```

```python
plt.figure(figsize=(12,6))
sns.histplot(df["Daily_Return"],bins=20,kde=True,color="Red")
plt.title("distribution of Daily_Return")
plt.xlabel("Daily_Return")
plt.ylabel("Frequency")
plt.xticks(rotation=45)
plt.show()
```
