```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
plt.figure(figsize=(12,6))
sns.regplot(
    data=df,
    x="sqft_living",
    y="price",
)

plt.title("distribution of sqft_living vs price")
plt.xlabel("sqft_living")
plt.ylabel("price")
plt.show()
```

```python
plt.figure(figsize=(12,6))
sns.boxplot(
    data=df,
    x="grade",
    y="price"
)

plt.title("distribution of grade vs price") 
plt.xlabel("grade")
plt.ylabel("price")
plt.show()
```

```python
plt.figure(figsize=(12,6))
sns.boxplot(
    data=df,
    x="waterfront",
    y="price"
)

plt.title("distribution of waterflot vs price")
plt.xlabel("waterfront")
plt.ylabel("price")
plt.show()
```

```python
plt.figure(figsize=(12,6))
sns.scatterplot(
    data=df,
    x="long",
    y="lat",
    hue="price",
    palette="Reds",
    alpha=0.6,
    legend=False
)

plt.title("longitude vs latitude")
plt.xlabel("longitutde")
plt.ylabel("latitude")
plt.show()
```

```python
location_summary=df.groupby("zipcode")["price"].agg(
    ["count","mean","median","max"]
).reset_index()
```

```python
location_summary=location_summary.sort_values(
    by="mean",
    ascending=False
)
location_summary.head(10)
```
