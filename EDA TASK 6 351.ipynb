```python
df["Stay of hospital"]=(df["Discharge Date"]-df["Date of Admission"]).dt.days
```

```python
df["Stay of hospital"]
```

```python
numeric_cols=["Age","Stay of hospital","Billing Amount"]
```

```python
correlation=df[numeric_cols].corr()
```

```python
correlation
```

```python
sns.heatmap(
    correlation,
    annot=True,
    cmap="YlGnBu",
    cbar=True,
    linewidth=3)
plt.title("Correlation matrix of age,stay duration and medical cost")
plt.show()
```
