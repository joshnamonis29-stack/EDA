```python
market_cancel=df.groupby("market_segment",as_index=False)["is_canceled"].mean()
market_cancel["cancellation_rate"]=market_cancel["is_canceled"]*100
print(market_cancel)
```

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
plt.figure(figsize=(10,6))
sns.barplot(
    data=market_cancel,
    x="market_segment",
    y="cncellation_rate"
)
plt.title("cancellation rate by market segment")
plt.xlabel("market segment")
plt.ylabel("cacellation rate")
plt.xticks(rotation=45)
plt.show()
```

```python
hotel_cancel=df.groupby("hotel",as_index=False)["is_canceled"].mean()
hotel_cancel["cancellation_rate"]=hotel_cancel["is_canceled"]*100
print(hotel_cancel)
```

```python
plt.figure(figsize=(10,6))
sns.barplot(
    data=market_cancel,
    x="hotel",
    y="cncellation_rate"
)
plt.title("cancellation rate by hotel type")
plt.xlabel("hotel type")
plt.ylabel("cacellation rate")
plt.xticks(rotation=45)
plt.show()
```

```python
month_order= [
    "Jan", "Feb", "Mar", "Apr", "May", "Jun", 
    "Jul", "Aug", "Sep", "Oct", "Nov", "Dec"
]
df["arrival_date_month"] = pd.Categorical(
    df["arrival_date_month"],
    categories=month_order,
    ordered=True
)
monthly_arrivals = df.groupby(
    "arrival_date_month",
    as_index=False,
    observed=True
).size()

monthly_arrivals.columns = ["month", "arrivals"]
```

```python
plt.figure(figsize=(12,5))

sns.lineplot(
    data=monthly_arrivals,
    x="month",
    y="arrivals",
    marker="o"
)
plt.title("Arrivals by Month")
plt.xlabel("Month")
plt.ylabel("Arrivals")
plt.xticks(rotation=45)

plt.show()
```

```python
monthly_adr=df.groupby(
    "arrival_date_month",
    as_index=False,
    observed=True
)["adr"].mean()

monthly_adr.columns=["month","adr"]
```

```python
monthly_adr
```

```python
plt.figure(figsize=(12,5))

sns.lineplot(
    data=monthly_adr,
    x="month",
    y="adr",
    marker="o"
)
plt.title("Monthly ADR Fluctuation")
plt.xlabel("Month")
plt.ylabel("Average Date Rate(%)")
plt.xticks(rotation=45)

plt.show()
```

```python
df["booking_status"] = df["is_canceled"].map({
    0: "Not Canceled",
    1: "Canceled"
})

plt.figure(figsize=(8, 5))

sns.boxplot(
    data=df,
    x="booking_status",
    y="lead_time"
)

plt.title("Lead Time Distribution")
plt.xlabel("Booking Status")
plt.ylabel("Lead Time (Days)")

plt.show()
```
