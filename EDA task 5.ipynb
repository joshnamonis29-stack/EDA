```python
def categorize_urgency(admission):
    if admission=="Emergency":
        return "Immediate care"
    elif admission=="Urgent":
        return "high priorty"
    else:
        return "planned care"
df["priority_level"]=df["Admission Type"].apply(categorize_urgency)
```
